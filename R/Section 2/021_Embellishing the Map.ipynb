### 2.2.3 Embellishing the map

You have now drawn your first maps in R showing the *roads*, *breaches of peace* and *census blocks* in **New Haven**. However, maps generally need more information than what was plotten in the previous page. In particular, someone who had no prior knowledge of **New Haven** may not realise the geographicla extent (in miles or kilometres) of the area. Assuming you kept the map window from the previous section, you can still add a scale using `map.scale` command. This command has a large number of parameters. Enter:

```r
plot(blocks, lwd = 0.5, border = "grey50")
plot(breach, col = "red", pch = 1, add = TRUE)
map.scale(534750, 152000, 15000, "km", 4, 0.5)
```

A scale bar is drawn (overplotted) on the map. the first two parameters are locatons at which the scalebar appears in the coordinate system of the map. The third parameter is the length (in projected map units) of the scale. In this case, it is 15000 km. The fourth parameter is a text string specifying the name of the units for the scale (km in this case). The fifth parameter gives the number of gradations in the scale, and the sixth gives the fraction of the units that each gradation on the scale represents. Here there are 5 gradations and each one represents 0.5 km.

A second embellishment is a north arrow. To add this, enter:

```r
plot(blocks, lwd = 0.5, border = "grey50")
plot(breach, col = "red", pch = 1, add = TRUE)
map.scale(534750, 152000, 15000, "km", 4, 0.5)
north.arrow(534750, 155000, 1000, col = "cyan")
```

This adds (overlays) the north arrow to the map. Again, the first two parameters are the location of the arrow. The thirds parameter specifies the size/length of the arrow and the `col` parameter specifies the colour that the arrow is filled with. if left unspecified, it will be filed in white. Remember that the `locator()` function introduced in Section 1 - this can be very useful for determining where to place items such as scale bars and north arrows in plot windows. A final decoration to the map is a map title. Suppose the map is to be called "New Haven, CT.". Then enter the code below to add the title to the map: 

```r
plot(blocks, lwd = 0.5, border = "grey50")
plot(breach, col = "red", pch = 1, add = TRUE)
map.scale(534750, 152000, 15000, "km", 4, 0.5)
north.arrow(534750, 155000, 1000, col = "cyan")
title('New Haven, CT.')
```

This looks more like a decent map now would you not say?

### 2.2.4 Saving your map

Having created a map in a window on the screen, you may now want to save the map for either printing, or incorporating in a document. There are several ways that this can be accomplished. The simplest way is to right-click with the mouse on the map window, select **copy** or **copy image to clipboard** (Mac) or **Copy as metafile** or **Copy as bitmap** (Windows), and then paste it into a word-processing document (openOffice or MS Windows). Another way is to highlight the window and to use **File > Save as** to save the map as an image file, with a name that you give it.
