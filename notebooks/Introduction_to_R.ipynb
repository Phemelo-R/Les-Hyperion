### **Data types**

In R there are various types of data, the most common four are **Numeric**, **Integer**, **Character** and **Factor**.

**Logical** and **Comples** are also data types however they are very rarely used explicitly.

- Numeric data is any real numbers so 8 or 12.3 or 1.00000002 etc.
- Integer data is just whole numbers 3, 4, 111 etc.
- Character data are words or letters surrounded by quotations (either ” or ‘) such as “A”, “Red”, ’Treated’, Character data has no order to it in Rs ‘mind’
- Factor data is like character data but r (or you) have assigned an order to it e.g. “A”, “B”, “C”.

````r
numeric_vector <- c(1.3,5.8,5.122,3.00,7.12)

# we change between data types with these functions
integer_vector <- as.integer(c(1, 5, 5, 3, 7)) 

character_vector <- c("krill", "adder", "fynbos")

logical_vector <- c(TRUE, FALSE, TRUE)

# Notice how r automatically orders alphabetically if we don't tell it the order
factor_vector <- as.factor(c("This","is","A","Character", "Vector"))
````

Check types:

````r
class(numeric_vector)
class(integer_vector)
class(character_vector)
class(logical_vector)
class(factor_vector)
````

Mixing types forces coercion:

````r
c(1, "a")
````

R will convert everything to character. It tries to maintain consistency within a vector.

We can also change between types (even if they don’t fit that description)

````r
Convert_numbers_to_characters <- as.character(numeric_vector)

Convert_numbers_to_characters
````

Now our numbers are thought of as characters, so we can’t apply numeric operations to them!

### Matrix

A matrix is a multi-dimensional data structure that can hold elements of the same data type (it could be Numeric, Character, Integer, Factor). A matrix can be created using the `matrix()` function in R.

````r
Matrix_numeric <- as.matrix(c(1.3,5.8,5.122,3.00,7.12))

Matrix_character <- as.matrix(c("This","is","A","Character Matrix"))
````

### DataFrame

A dataframe is the most commonly used 'container' for data in R. A dataframe is a combination of multiple groups of data types (it could be Numeric, Character, Integer, Factor), with two dimensions. Each element of a dataframe must have the same length as the other element.

````r
ecology_data <- data.frame(
  species = c("krill", "adder", "proteaceae"),
  abundance = c(120, 15, 45),
  site = c("A", "B", "C")
)
````

View it

````r
ecology_data
````

Access columns

````r
ecology_data$species
ecology_data$abundance
````

Check structure

````r
str(ecology_data)
summary(ecology_data)
````

### Lists

Lists are multiple groups of a combination of data types or objects (it could be Numeric, Character, Integer, Factor or vectors, dataframes or matrices of these), with two dimensions. Each element in a list can be a different length to the other elements. Lists are useful for storing complex data structures, can also be accessed using indexing.

````r
List_numeric <- list(c(1.3, 5.8, 5.122, 3.00, 7.12),
                   c(1, 5, 5, 3, 7))

List_from_vectors <- list(character_vector, Matrix_numeric, Matrix_numeric)

List_different_lengths <- list(Item1 = c(1, 2, 3, 4, 5, 6), Item2 = c("a", "B", "C", "D"), Item3 = seq(from = 1, to = 100, by = 1))

List_different_lengths
````

### Array

An array refers to multi-dimensional data structures that can hold elements of the same data type. It can have any number of dimensions and its dimensions can be named. **Multi-dimensional arrays are common in modelling and spatial data.**

````r
Array_1d <- array(c(Matrix_numeric,c(1.3,5.8,5.122,3.00,7.12)),dim=c(5))
Array_2d <- array(c(Matrix_numeric,c(1.3,5.8,5.122,3.00,7.12)),dim=c(5,2))
Array_3d <- array(c(Matrix_numeric,c(1.3,5.8,5.122,3.00,7.12)),dim=c(5,2,2))
````

### **Indexing**

Object have dimensions and there is a technique we can use called **indexing** to select elements of an object. We often use square brackets to do this. If the object is 1 dimensional, one number will return one value.

````r
numeric_vector[4]
````

If the object is 2 dimensional, one number will return that column

````r
ecology_data[2]
````

By adding a comma we can select the from both dimensions (rows first, then columns)

````r
ecology_data[3, 2]
````

If we want all rows but only a specific column we add a comma without a number

````r
ecology_data[, 2]
````

And vice verse

````r
ecology_data[3,]
````

### Data inspection

First we can make some data into a dataframe, explore this data, then save it as a file and then read the file back into R. R has some very useful random and non-random data generation functions

````r
Year <- seq(from = 1950, to = 2023, by = 1)
Treatment <- c("Control", "Treatment 1", "Treatment 2")
Rep <- seq(from = 1, to = 10, by = 1)
````

These are three vectors, which we can check information about them with a few simple functions

````r
length(Year)
````

````r
summary(Year)
````

````r
length(Treatment)
````

````r
summary(Treatment)
````

````r
length(Rep)
````

````r
summary(Rep)
````

We want to combine these vectors so we have a row for each **Rep**, **Year** and **Treatment**, we can do this by expanding the grid and create a new dataframe called **df**. We can inspect specific elements of a dataframe too.

````r
df <- expand.grid(Year = Year, Treatment = Treatment, Rep = Rep)

class(df) # Type of object
````

````r
nrow(df) # Number of rows
````

````r
ncol(df) # Number of columns
````

````r
dim(df) # Dimensions of object
````

````r
head(df) # The first 6 rows of the df
````

````r
tail(df) # The last 6 rows of the df
````

This **df** is all the metadata we want for our dataframe that we want to now make up some response data

````r
# We need the response to be same length as the df so we use nrow() for the number of values we want.
Response <- rnorm(n = nrow(df), mean = 15, sd = 8) 
````

We can then combine this to our **df**, the dollar sign ($) is used to select one dimension (column) from within an object (here a dataframe). The column Response isn’t present in the data but by assigning our Response vector to it with `df$Response <- Response` it adds a new column called Response to the dataframe df.

````r
df$Response <- Response
````

### **Data saving**

We now can save the **df** we created with the Resoponce column to a new folder using the `write.csv()` function from base R, or even better the `write_csv()` function from the `readr` package.

To save to our folder we only need to say the directory we want the file saved to, followed by a '/', then the name of the file with file extension. Inside reading and writing functions such as `write_csv()` the main argument will be where is the file to be saved to or taken from and we write this out as a character string inside quotations.

````r
write_csv(df,"NewFolderName/OurNewFile.csv")
````
