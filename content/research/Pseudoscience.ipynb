# Evaluating Protex Deep Clean Bar Soap: Scientific or Pseudoscientific?

## Introduction

Protex is an established brand of antibacterial soaps owned by Colgate-Palmolive, with formulations that differ depending on the market. One of their soaps is Protex Deep Clean Bar Soap, which is a widely marketed antibacterial soap that claims to offer “superior protection against germs” compared to regular soap and even suggests it can help with skin problems like acne and dark spots (Colgate-Palmolive Company, 2017). These claims are printed on the packaging and shown in advertisements, often with images of a flaxseed oil drop on the epidermis under a microscope or phrases like “protects against 99.9% of germs”. Scientific studies and regulatory health agencies have questioned whether antibacterial soaps like Protex truly work better than plain soap and water (Aiello et al., 2007; Kim et al., 2015). 

This study aims to evaluate Protex Deep Clean Bar Soap’s claims and see whether they are scientifically true, or they rely on pseudoscience marketing tactics to gain sales. In this study, we will assess the bar soap’s ingredients, the legitimacy of its antibacterial claims, and whether their approach in making the product aligns with scientific principles.

## The science behind Protex Deep Clean Bar Soap

Protex Deep Clean Bar Soap originally contained active antibacterial agents such as triclosan or triclocarban, which are chemicals that are known for their antimicrobial properties and have been used in antibacterial soaps for years (Giuliano and Rybak, 2015). However, in 2016, the U.S. Food and Drug Administration (FDA) banned them from consumer soaps. The FDA banned them because they lack sufficient evidence to prove that they are safer to use or work better than plain soap and water, as well as some concerns about long-term health and environmental effects (FDA, 2016).

After the ban, Protex reformulated its soap with other antibacterial chemicals, such as benzalkonium chloride (BAC) and chloroxylenol (Pereira and Tagkopoulos, 2019). These chemicals work by disrupting bacterial cell membranes and interfering with essential enzymatic processes in microorganisms, therefore killing or inhibiting bacterial growth (Pereira and Tagkopoulos, 2019). Laboratory studies confirm that these chemicals can kill bacteria in controlled experiments (Pereira and Tagkopoulos, 2019). However, killing bacteria in a lab dish does not automatically mean that the soap works better during real-life handwashing. 

## Evidence Analysis: Lab Results vs. Real-World Use

The biggest problem with Protex’s claims is the gap between lab tests and everyday use. Multiple peer-reviewed studies have shown that antibacterial soaps containing triclosan do not prevent illness or remove germs better than regular soap in real-world scenarios. For example, a study published in Clinical Infectious Diseases followed over 1 000 households for one year. Half the households used antibacterial soap with triclosan, while the other half used plain soap. The results showed no difference in rates of colds, flu, or skin infections between the two groups (Aiello et al., 2007). In a study that was published in the Journal of Antimicrobial Chemotherapy, researchers at the University of Seoul compared triclosan soap to regular soap. In lab tests, triclosan killed bacteria after 9 minutes of contact. However, when volunteers washed their hands for 30 seconds, which was the time doctors recommended, both soaps removed germs equally well. The act of physically scrubbing and rinsing, not the antibacterial chemicals, did most of the work (Kim et al., 2015). In 2016, the FDA came to a conclusion that there was no proof that antibacterial soaps prevent illness more effectively than regular soap. They also warned that long-term use of antibacterial chemicals like triclosan could lead to antibiotic resistance (FDA, 2016).
The new ingredients protex use, like BAC, also face the same issues. For BAC to kill bacteria, it needs 2–6 minutes of contact time (Pereira and Tagkopoulos, 2019). Most people only wash their hands for 20 seconds, which is far too short for the chemicals to work. This means that the antibacterial agents in Protex are likely rinsed down the drain before they can even do anything.

## Methodology Evaluation: Flawed Testing Practices

The methodology behind Protex's claims fail to meet the strict scientific standards. Even if the manufacturers may have conducted lab tests showing bacterial reduction, these controlled experiments do not represent how people actually use soap. Protex has not conducted large-scale studies to prove its soap reduces infections or improves skin health. Instead, they cite lab data about killing bacteria in dishes. This is like claiming a car can fly because it has wings in a video game, it completely ignores how things work in reality. For proper scientific validation, they need large-scale clinical trials that will actually show reduced bacterial infection rates among the users (ATSDR, 2007), which is something that is not there from the product's supporting evidence. 

## Red Flags of Pseudoscience

Several red flags characteristic of pseudoscience appear in Protex's marketing strategy. The “protects against 99.9% of germs” claim sounds impressive but is based on lab conditions that don’t match real use. It also avoids mentioning which specific germs are targeted. Advertisements show flaxseed oils entering the skin and killing germs, people in lab coats, and slogans like “advanced protection.” These advertisements create a false impression of scientific authority without providing any real evidence (Hartley, 2021). Testimonials of people are unreliable because they don’t take into account other factors such as diet and skincare routines. When scientists actually point out that regular soap works just as well, Protex does not address these criticisms, instead, they repeat statements like “our formula is clinically proven” without sharing any details.

## Conclusion: Science or Clever Marketing?

In conclusion, while Protex Deep Clean Bar Soap contains ingredients with scientifically verified antibacterial properties, its marketing exaggerates the benefits for everyday use and that makes it lean more towards pseudoscience. Scientific evidence shows that Antibacterial soaps are no better than regular soap at preventing illness. The chemicals in Protex need minutes to work, which is far longer than a typical handwash, and overuse of antibacterial products may contribute to antibiotic-resistant bacteria, making infections harder to treat. The World Health Organization (WHO) recommends washing hands with plain soap and water for 20 seconds. There is no need for antibacterial additives. Protex’s claims are a mix of half-truths and marketing strategies, making it more pseudoscience than real science.
